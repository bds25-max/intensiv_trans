# Русская транскрибация аудио и видео

Запустите ячейку ниже, выберите аудио- или видеофайл. После распознавания текст появится под ячейкой и скачается в формате `.txt`.


```python
!pip -q install openai-whisper

from google.colab import files
from pathlib import Path
import whisper

uploaded = files.upload()
filename = next(iter(uploaded))

model = whisper.load_model("small")
result = model.transcribe(filename, language="ru", task="transcribe")
text = result["text"]

print(text)

txt_name = str(Path(filename).with_suffix(".txt"))
Path(txt_name).write_text(text, encoding="utf-8")
files.download(txt_name)
```
